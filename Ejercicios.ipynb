Ejercicio:
Integra: $$f(x) = x^{2} + 3x + 9$$

Con el metodo del trapecio: $$\int_a^b f(x) dx \approx \frac{\Delta x}{2} \left[ f(x_0) + 2 \sum_{i=1}^{n-1} f(x_i) + f(x_n) \right]$$

```python
# Parámetros iniciales de la integración
a = -2.0  
b = 3.0     
n = 1000    

# Calculamos el ancho de cada subintervalo (h)
h = (b - a) / n
```

```python
# Evaluamos los extremos f(a) y f(b)
f_a = (a**2) + (3 * a) + 9
f_b = (b**2) + (3 * b) + 9
```

```python
# calculamos inicial con la suma con los valores de los extremos
suma_total = f_a + f_b

for i in range(1, n):
    x = a + (i * h)
    # Evaluación de f(x) en el punto intermedio
    f_x = (x**2) + (3 * x) + 9
    # Se multiplica por 2 según la regla del trapecio
    suma_total += 2 * f_x
    
area = (h / 2) * suma_total

print(f"La aproximación de la integral es: {area}")
```
