1. La constante de Madelung

En física de la materia condensada, la constante de Madelung da el potencial eléctrico total que siente un átomo en un sólido; y depende de las cargas de los otros átomos cercanos y de sus ubicaciones. Por ejemplo, el cristal de cloruro de sodio sólido (la sal de mesa), tiene átomos dispuestos en una red cúbica, con átomos de sodio y cloro alternados, teniendo los de sodio una carga positiva $+e$ y los de cloro una negativa $-e$, (donde $e$ es la carga del electrón). Si etiquetamos cada posición en la red con tres coordenadas enteras $(i,j,k)$, entonces los átomos de sodio caen en posiciones donde $i+j+k$ es par, y los átomos de cloro en posiciones donde $i+j+k$ es impar.

Consideremos un átomo de sodio en el origen, i.e. $i = j = k = 0$, y calculemos la constante de Madelung. Si el espaciado de los átomos en la red es $a$, entonces la distancia desde el origen al átomo en la posición $(i,j,k)$ es:

$$ \sqrt{(ia)^2 + (ja)^2 + (ka)^2} = a\sqrt{i^2 + j^2 + k^2}$$

y el potencial en el origen creado por tal átomo es:

$$V(i,j,k) = \pm \frac{e}{4\pi\epsilon_0 a \sqrt{i^2 + j^2 + k^2}}$$

siendo $\epsilon_0$ la permitividad del vacío y el signo de la expresión se toma dependiendo de si $i+j+k$ es par o impar. Así entonces, el potencial total que siente el átomo de sodio es la suma de esta cantidad sobre todos los demás átomos. Supongamos una caja cúbica alrededor del átomo de sodio en el origen, con $L$ átomos en todas las direcciones, entonces:

$$V_{\text{total}} = \sum_{\substack{i,j,k=-L \\ i,j,k \neq 0}}^{L} V(i,j,k) = \frac{e}{4\pi\epsilon_0 a} M$$

donde $M$ es la constante de Madelung (al menos aproximadamente).

Técnicamente, la constante de Madelung es el valor de $M$ cuando $L \to \infty$, pero se puede obtener una buena aproximación simplemente usando un valor grande de $L$. Escribe un programa para calcular e imprimir la constante de Madelung para el cloruro de sodio. Utiliza un valor de $L$ tan grande como puedas, sin dejar que tu programa se ejecute en un tiempo razonable (un minuto o menos).

Solución:
Primero definimos las variables que nunca van a cambiar:

```python
import numpy as np
e = 1.602e-19 #Coulombs
E_0 = 8.854e-12 #F/m
a = 2.82e-10 #m (Valor buscado en internet)
pi = np.pi
```

Ahora sacamos las constantes de la serie: $$V_{\text{total}} = \frac{e}{4\pi \epsilon_{0} a} \sum_{\substack{i,j,k=-L \\ i,j,k \neq 0}}^{L}$$

De modo que solo nos quedaria resolver: $$M = \sum_{\substack{i,j,k=-L \\ i,j,k \neq 0}}^{L} \frac{\pm 1}{\sqrt{i^2 + j^2 + k^2}}$$

No era necesario establecer las constantes jeje

```python
#Definimos el valor de i,j,k,

L = 250  #El valor de L para la iteracion
```

Ahora programemos la suma:

```python

M = 0 #La variable que ira guardando los valores
for i in range(-L,L+1):
    for j in range(-L,L+1):
        for k in range(-L,L+1):
            if i == 0 and j == 0 and k == 0:#Este es para evitar que todos los indices se hagan 0 y se indetermine
                continue
            #Calculamos la norma de nuestros indices
            norm = np.sqrt((i)**(2) + (j)**(2) + (k)**(2))
            #calculamos la suma de los indices
            suma = i + j + k
            #analizamos si es par o impar y con base a eso actualizamos a m
            p = suma%2
            if p == 0:
                M += 1/(norm) #Si es par se suma
            else:
                M -= 1/(norm) #Si no es par se resta
            
        
print(f"La constante de Madelung es aproximadamente: {M}")
```

En particular la fórmula se usa para calcular la *energía de enlace nuclear aproximada* $B$, de un núcleo atómico con número atómico $Z$ y número de masa $A$:

$$B = a_1 A - a_2 A^{2/3} - a_3 \frac{Z^2}{A^{1/3}} - a_4 \frac{(A-2Z)^2}{A} + \frac{a_5}{A^{1/2}},$$

donde, en unidades de millones de electrón-volts, las constantes son $a_1 = 15.8, a_2 = 18.3, a_3 = 0.714, a_4 = 23.2$ y

$$
a_5 = \begin{cases} 
0 & \text{si } A \text{ es impar,} \\ 
12.0 & \text{si } A \text{ y } Z \text{ son pares (ambos),} \\ 
-12.0 & \text{si } A \text{ es par y } Z \text{ impar.} 
\end{cases}
$$

### **4. Paso de potencial cuántico.** 
#### Un conocido problema de mecánica cuántica involucra una partícula de masa $m$ que encuentra un paso de potencial unidimensional, como este:

<center><img src="qstep2.png" width="50%" class="center"/></center>

#### La partícula con energía cinética inicial $E$ y vector de onda $k_1=\sqrt{2mE}/\hbar$ entra por la izquierda y encuentra un salto repentino en la energía potencial de altura $V$ en la posición $x=0$. Resolviendo la ecuación de Schrödinger, se puede demostrar que cuando $E>V$ la partícula puede 

#### (a) **pasar el escalón**, en cuyo caso tiene una energía cinética menor a $E-V$ en el otro lado y un vector de onda correspondientemente menor a $k_2=\sqrt{2m(E-V)}/\hbar$, \; o bien (b) **reflejarse**, manteniendo toda su energía cinética y un vector de onda sin cambios, pero moviéndose en la dirección opuesta. Las probabilidades $T$ y $R$ de transmisión y reflexión están dadas por:

<h3 align="center">$T = {4k_1k_2\over(k_1+k_2)^2}\,,\qquad$
$R = \biggl( {k_1-k_2\over k_1+k_2} \biggr)^2.$</h3>

#### Supongamos que tenemos una partícula con una masa igual a la masa del electrón. $m=9.11\times10^{-31}\,$kg y energía de $10\,$eV, al encontrar un escalón de potencial de altura $9\,$eV. Escribe un programa en \texttt{Python} para calcular e imprimir las probabilidades de transmisión y reflexión utilizando las fórmulas anteriores.


```python
#Primero definimos el valor de cada variable que va a ser constante
#Colocamos el valor de h barra en eV
ℏ = 6.58211e-16
#La masa
m = 9.11e-31
#La energia en eV
E = 10
#El potencial en eV
V = 9
```

```python
#Ahora calculamos el valor de k1 y k2
k1 = (mt.sqrt(2*m*E))/ℏ
k2 = (mt.sqrt(2*m*(E-V)))/ℏ
```

```python
#Por ultimo le pedimos que nos pida si queremos ver las probabilidades
while (True):
    p = str(input("Coloque ':D' si gusta la probabilidad de transmitancia y reflectancia: "))
    if (p == ':D'):
        T = (4*k1*k2)/((k1+k2)**2)
        R = ((k1-k2)/(k1+k2))**2
        print(f"La probabilidad de transmitancia es: {T}")
        print(f"La probabilidad de reflectancia es: {R}")
        break
    else:
        print("No quiere las probabilidades??")
        print("Ta bien le doy otra oportunidad")
```
